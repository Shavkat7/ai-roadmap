# 1. Preprocessing #

```python
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns
numerical_features
```

```python
# Identyfying binary (boolean - yes/no) columns

binary_columns = [col for col in df.columns if df[col].nunique() == 2]
binary_columns
```

```python
df[binary_columns] = df[binary_columns].applymap(lambda x: 1 if x == 'yes' else 0)
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

on_h_en = OneHotEncoder(sparse_output=False)

on_h_en.fit(df[['furnishingstatus']])

encoded_df_furnishingstatus = pd.DataFrame(data=on_h_en.transform(df[['furnishingstatus']]), columns=on_h_en.get_feature_names_out())

df = df.drop(['furnishingstatus'], axis=1)
df = pd.concat([df, encoded_df_furnishingstatus], axis=1)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import MinMaxScaler

min_max_scaler = MinMaxScaler()

df[numerical_features] = min_max_scaler.fit_transform(df[numerical_features])
```

```python
df.head()
```

```python
# Splitting into train-test

from sklearn.model_selection import train_test_split

X = df.drop('price', axis=1)
Y = df['price']

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression

mult_lin_reg = LinearRegression()

mult_lin_reg.fit(x_train, y_train)

```

```python
mult_lin_reg.intercept_ , mult_lin_reg.coef_
```
